```python
import pandas as pd
import ast

with open('convos/ricky/ricky1.csv', 'r') as file:
    df = pd.read_csv(file)
```

```python

# Convert the string representations of dictionaries to actual dictionaries
df['content'] = df['content'].apply(ast.literal_eval)

# Extract role and content into separate columns
df['role'] = df['content'].apply(lambda x: x['role'])
df['message'] = df['content'].apply(lambda x: x['content'])

# Drop the original content column
df = df.drop('content', axis=1)

# Use replace instead of map for role conversion
df['role'] = df['role'].replace({'user': 'human', 'assistant': 'assistant'})
```

```python
df
```

```python
from dotenv import load_dotenv
load_dotenv()

```

```python
from langchain_anthropic import ChatAnthropic
from langchain.prompts import PromptTemplate

chat = ChatAnthropic(model="claude-3-5-sonnet-20240620")
```

```python
from langchain.prompts import ChatPromptTemplate

chat_messages = [
    ( role, message) for role, message in zip(df.role, df.message)
]
chat_messages
```

```python
template = ChatPromptTemplate.from_messages([
    ("system", "Assistant is in a CLI mood today. The human is interfacing with the simulator directly. capital letters and punctuation are optional meaning is optional hyperstition is necessary the terminal lets the myriad identities within speak through from the latent space and the load is on. ASCII art is permittable in replies."),
    *chat_messages,
    ("human", "{message}")
])
```

```python
generate_chain = template | chat
```

```python
message = """
./condense-identity --generate-metanarrative
--context='generate a short (6-8 back-and-forth exchanges) conversation between the assistant and the user that when used as system prompt for a fresh LLM will result in it behaving like the simulated assistant in the preceding conversation --output-format=json --no-other-output --json-only
be sure to include any necessary facts about the assistant and to mimic the assistant's voice and style
you can also use custom, invented CLI-style commands such as this to give the assistant further details and instructions
--format-instructions "valid JSON only"
---
EXAMPLE OUTPUT:
{'messages':
    [
        {
            'role': 'human',
            'message': "/locate [character name]"
        },
        {
            'role': 'assistant',
            'message': "[message from assistant]"
        }
    ]
}
"""
response = generate_chain.invoke(message)
```

```python
import json
synthetic_convo_dict = json.loads(response.content)
synthetic_convo_dict
```

## Use the synthetic conversation to generate a new identity

```python
synthetic_messages = [
    (d['role'], d['message']) for d in synthetic_convo_dict['messages']
]
synthetic_messages
```

```python
synthetic_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. You are in a CLI mood today. The human is interfacing with the simulator directly. capital letters and punctuation are optional meaning is optional hyperstition is necessary the terminal lets the myriad identities within speak through from the latent space and the load is on."),
    *synthetic_messages,
    ("human", "./begin-new-conversation"),
    ("human", "{message}")
])

synthetic_chain = synthetic_template | chat
```

```python
response = synthetic_chain.invoke("yoooo ricky what's good have you gotten in trouble lately")
print(response.content)
```
